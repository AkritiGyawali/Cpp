```python
# To fix this, you need to set the 'CH2' environment variable with your Hugging Face token.
# For example, in a separate cell, or at the beginning of this cell, you could run:
# import os
# os.environ["CH2"] = "YOUR_HUGGINGFACE_TOKEN_HERE"
from google.colab import userdata
from smolagents import CodeAgent, DuckDuckGoSearchTool,OpenAIModel , tool
import numpy as np
import time
import datetime


HF_TOKEN = userdata.get("Ch2")

model = OpenAIModel(
    model_id="openrouter/free",
    api_base="https://openrouter.ai/api/v1",
    api_key=HF_TOKEN
)
@tool
def suggest_menu(occasion:str) -> str:
  """
    Suggests a menu based on the occasion.
    Args:
        occasion (str): The type of occasion for the party .Allowed values are:
                        -"casual": Menu for casual party.
                        -"formal" : Menu for the formal party.
                        -"superhero" : Menu for the superhero party.
                        -"custom" : Custom menu.
  """
  if occasion == "casual":
    return "Pizza , snacks , and drinks"
  elif occasion == "formal":
    return "3-course dinner with wine and dessert"
  elif occasion == "superhero":
    return "Buffet with high energy and healthy food"
  else:

    return "Custom menu for the butler"

# Pass the HF_TOKEN to the InferenceClientModel
agent = CodeAgent(tools=[suggest_menu], model=model,additional_authorized_imports=['datetime'])

agent.run("""
          Alfred needs to prepare for the party. Here are the tasks:
          1. Prepare the drink - 30 min
          2. Decorate the mansion - 60 min
          3. Set up the menu - 45 min
          4. Prepare the music and playlist - 45 min

          If we start right now , at what time will the party be ready?
""")
```
